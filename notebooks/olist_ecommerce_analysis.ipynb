```python
import pandas as pd
orders = pd.read_csv("olist_orders_dataset.csv")
print(orders.head())
orders["order_purchase_timestamp"] = pd.to_datetime(
    orders["order_purchase_timestamp"]
    )
print(
 orders["order_purchase_timestamp"].dt.year.head()
 )
print(
 orders["order_purchase_timestamp"].dt.month.head()
)
status = (orders.groupby("order_status").agg ({
"order_id":"count"
})
)
customers = pd.read_csv("olist_customers_dataset.csv")

orders_customers = pd.merge(orders, customers, on="customer_id")
city_orders = (orders_customers.groupby("customer_city").agg ({"order_id":"count"}))
city_orders = city_orders.sort_values(by="order_id", ascending=False)

order_items = pd.read_csv("olist_order_items_dataset.csv")

products = pd.read_csv("olist_products_dataset.csv")

orders_items_products = pd.merge(order_items , products, on="product_id")

payments = pd.read_csv("olist_order_payments_dataset.csv")

payment_type = (payments.groupby("payment_type").agg({"order_id":"count"}).sort_values(by="order_id", ascending=False))

reviews = pd.read_csv("olist_order_reviews_dataset.csv")

rating = reviews.groupby("review_score").agg({"review_id":"count"})
```

```python
#4.Visualization
import matplotlib.pyplot as plt
# 1. Order Status
status_plot = status.sort_values("order_id", ascending=False)

plt.figure(figsize=(10, 5))
plt.bar(status_plot.index, status_plot["order_id"])

plt.title("Jumlah Order Berdasarkan Status")
plt.xlabel("Order Status")
plt.ylabel("Jumlah Order")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
# 2. Top 10 Kota
top_cities = city_orders.head(10)

plt.figure(figsize=(10, 5))
plt.bar(top_cities.index, top_cities["order_id"])

plt.title("Top 10 Kota Berdasarkan Jumlah Order")
plt.xlabel("Kota")
plt.ylabel("Jumlah Order")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
# 3. Top 10 Produk
top_products = (
    orders_items_products
    .groupby("product_id")
    .agg({"order_id": "count"})
    .sort_values("order_id", ascending=False)
    .head(10)
)

plt.figure(figsize=(10, 5))
plt.bar(top_products.index.astype(str), top_products["order_id"])

plt.title("Top 10 Produk Berdasarkan Jumlah Order")
plt.xlabel("Product ID")
plt.ylabel("Jumlah Order")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
# 4. Payment Type
payment_plot = payment_type.sort_values("order_id", ascending=False)

plt.figure(figsize=(8, 5))
plt.bar(payment_plot.index, payment_plot["order_id"])

plt.title("Jumlah Order Berdasarkan Metode Pembayaran")
plt.xlabel("Metode Pembayaran")
plt.ylabel("Jumlah Order")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
# 5. Customer Rating
rating_plot = rating.sort_index()

plt.figure(figsize=(8, 5))
plt.bar(rating_plot.index.astype(str), rating_plot["review_id"])

plt.title("Distribusi Rating Pelanggan")
plt.xlabel("Rating")
plt.ylabel("Jumlah Review")

plt.tight_layout()
plt.show()
```
